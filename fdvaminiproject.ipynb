```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset
data = {
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Maths":[78,65,90,55,88,72,60,95,81,69],
    "Python":[85,70,92,60,90,75,65,96,84,73],
    "DBMS":[80,68,88,58,85,78,62,94,79,71],
    "Attendance":[92,85,96,72,90,88,76,98,91,82]
}

df = pd.DataFrame(data)

# Average marks
df["Average"] = df[["Maths","Python","DBMS"]].mean(axis=1)

print("Student Performance Data:\n")
print(df)
print("\nSummary:\n", df.describe())

# Visualization
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.bar(df["Student"], df["Average"])
plt.title("Average Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.subplot(2,2,2)
df[["Maths","Python","DBMS"]].mean().plot(kind="bar")
plt.title("Subject-wise Average")
plt.ylabel("Marks")

plt.subplot(2,2,3)
plt.hist(df["Attendance"], bins=5)
plt.title("Attendance Distribution")
plt.xlabel("Attendance %")

plt.subplot(2,2,4)
sns.scatterplot(x="Attendance", y="Average", data=df)
plt.title("Attendance vs Average")

plt.tight_layout()
plt.show()
```
