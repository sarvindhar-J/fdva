```python
import pandas as pd

# Sample dataset
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Eshan'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 78, 92, 88, 76]
}

df = pd.DataFrame(data)

# Display data
print("Original Data:")
print(df)

# Export to CSV
df.to_csv('/content/student_data.csv', index=False)

# Export to Excel
df.to_excel('/content/student_data.xlsx', index=False)

# Import CSV
df_csv = pd.read_csv('/content/student_data.csv')
print("\nCSV DATA:")
print(df_csv.head())

# Import Excel
df_excel = pd.read_excel('/content/student_data.xlsx')
print("\nEXCEL DATA:")
print(df_excel.head())
```
